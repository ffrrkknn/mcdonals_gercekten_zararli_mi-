```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Grafiklerin düzgün görünmesi için stil ayarı
sns.set_theme(style="whitegrid")

# McDonald's en popüler ürünleri ve Ev Yapımı Sağlıklı Alternatifler verisi
data = {
    'Yiyecek': [
        'Big Mac', 'McDouble', 'Büyük Patates', 'McNuggets (9lu)', 'Kula (Büyük)',
        'Izgara Tavuklu Salata', 'Ev Yapımı Köfte Burger', 'Fırın Patates (Zeytinyağlı)', 'Ev Yapımı Tavuk Sote', 'Taze Sıkılmış Portakal Suyu'
    ],
    'Kategori': [
        'McDonalds', 'McDonalds', 'McDonalds', 'McDonalds', 'McDonalds',
        'Sağlıklı', 'Sağlıklı', 'Sağlıklı', 'Sağlıklı', 'Sağlıklı'
    ],
    'Kalori (kcal)': [540, 390, 510, 400, 290, 220, 350, 160, 250, 110],
    'Toplam Yağ (g)': [28, 19, 24, 24, 0, 5, 12, 4, 7, 0],
    'Tuz / Sodyum (mg)': [1010, 920, 350, 750, 10, 480, 250, 50, 180, 5],
    'Şeker (g)': [9, 7, 0, 1, 75, 4, 3, 1, 2, 20],
    'Protein (g)': [25, 20, 6, 22, 0, 28, 26, 3, 30, 2]
}

df = pd.DataFrame(data)
print("--- Veri Tablosu Oluşturuldu ---")
display(df)
```

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=df, x='Yiyecek', y='Kalori (kcal)', hue='Kategori', palette={'McDonalds': 'red', 'Sağlıklı': 'green'})
plt.xticks(rotation=45, ha='right')
plt.title('McDonald\'s vs Sağlıklı Yemekler: Kalori Karşılaştırması', fontsize=14)
plt.xlabel('')
plt.ylabel('Kalori (kcal)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=df, x='Yiyecek', y='Tuz / Sodyum (mg)', hue='Kategori', palette={'McDonalds': 'red', 'Sağlıklı': 'green'})
plt.axhline(2300, color='darkred', linestyle='--', label='Günlük Maksimum Tuz Sınırı (2300 mg)')
plt.xticks(rotation=45, ha='right')
plt.title('Gizli Tehlike: Tuz (Sodyum) Miktarları', fontsize=14)
plt.xlabel('')
plt.ylabel('Sodyum (mg)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=df, x='Yiyecek', y='Toplam Yağ (g)', hue='Kategori', palette={'McDonalds': 'darkred', 'Sağlıklı': 'forestgreen'})
plt.xticks(rotation=45, ha='right')
plt.title('Yiyeceklerdeki Yağ Miktarı (Gram)', fontsize=14)
plt.xlabel('')
plt.ylabel('Yağ (g)')
plt.tight_layout()
plt.show()
```

# 🍎 McDonald's Gercekten Zararli mi?

### 🎯 Maddelerin Vucudumuza Etkisi
* **Protein (Faydali):** Kaslarimizi guclendirir. Hem fast-food hem ev yemeginde bulunur.
* **Kalori (Dengeli Olmali):** Gunluk enerjimizdir. Fazlasi vucutta yag olarak birikir.
* **Tuz ve Yag (Fazlasi Zararli):** Kalbi yorar, kilo aldirir ve damar sagligini bozar.
* **Seker (Zararli):** Yuksek miktarda seker ani acikmaya ve kilo alimina yol acar.

---

### 🔍 Analiz Sonuclari
1. **Asiri Tuz ve Kalori:** Bir Big Mac, buyuk patates ve buyuk kola toplam **1340 kalori** ve **1360 mg tuz** icerir. Bu, gunluk tuz limitinin yaridan fazlasini tek ogunde bitirir.
2. **Gizli Seker Deposu:** 1 buyuk boy kolada **75 gram seker** (yaklasik 18 kup seker) vardir.
3. **Ev Yemegi Farki:** Ev yapimi kofte burger veya tavuk sote, ayni proteini yari yariya daha az yag ve tuz ile saglar.

---

### 📌 Ozet
McDonald's yemekleri yuksek protein saglasa da icerdigi **asiri tuz, doymus yag ve gizli seker** nedeniyle sik tuketildiginde sagliga zararlidir. Ev yapimi alternatifler vucudu yormadan ayni besin degerini verir.
